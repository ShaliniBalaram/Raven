# RAVEN Walkthrough

This notebook mirrors the CLI pipeline while keeping each intermediate object visible for inspection.

```python
from raven import Pipeline

pipe = Pipeline.from_yaml('../configs/example_aoi.yml')
pipe.config
```

```python
dem = pipe.load_dem()
conditioned = pipe.condition_dem(dem)
fdir = pipe.flow_direction(conditioned)
facc = pipe.flow_accumulation(fdir)
streams = pipe.extract_streams(facc)
basins = pipe.delineate_basins(fdir, streams)
reach_graph = pipe.build_graph(streams, basins)

len(streams.reaches), len(basins.basins), reach_graph.number_of_nodes()
```
